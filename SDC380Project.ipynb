```python
from datetime import datetime
import pandas as pd
from IPython.display import display, Markdown
# Load the dataset into a new DataFrame

file_path = 'StudentsPerformance.xlsx'
df = pd.read_excel(file_path)

import pandas as pd #import pandas library

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "johhun3213"
display(Markdown(f"### {student_id} - {todays_date}")) #display header text

mean_math_score = df['math score'].mean() #Mean math scores
std_math_score = df['math score'].std() #Standard Deviation
median_math_score = df['math score'].median() #Median math scores

min_math_score = df['math score'].min()
max_math_score = df['math score'].max()

print("Mean Math Scores: ", mean_math_score)
print("Standard Deviation of Math Scores: ", std_math_score)
print("Median Math Scores: ", median_math_score)
print("Lowest Math Score: ", min_math_score)
print("Highest Math Score: ", max_math_score)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from datetime import datetime

student_id = "johhun3213"

pd.read_excel('StudentsPerformance.xlsx')

current_datetime = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
plt.figure(figsize=(10, 6))
plt.hist(df['math score'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.title(f'Histogram of Math Scores for Students - StudentID: {student_id} - {current_datetime}')
plt.grid(True)
plt.show()
```

```python
avg_math = df.groupby("race/ethnicity")["math score"].mean().round(2)

plt.figure(figsize=(8, 5))
bars = plt.barh(avg_math.index, avg_math.values, color="green", edgecolor="black")
plt.bar_label(bars, padding=3)  # puts the average on top of each bar
plt.title(f'Average Math Scores by Race/Ethnicity - STUDENTID: {student_id} - {current_datetime}')
plt.xlabel("Average Math Score")
plt.ylabel("Race/Ethnic Group")
plt.gca().invert_yaxis()
plt.grid(True)
plt.show()
```

```python
x = df['math score']
y = df['writing score']
z = np.polyfit(x, y, 1)
p = np.poly1d(z)
plt.figure(figsize=(12, 8))
plt.scatter(x, y, label='Data Points', color='blue')
plt.plot(x, p(x), label=f'Trendline: y = {z[0]:.2f}x + {z[1]:.2f}', color='red')
plt.xlabel('Math Scores')
plt.ylabel('Writing Scores')
plt.title(f'Relationship between Math Scores and Writing Scores - StudentID: {student_id} - {current_datetime}')
plt.legend()
plt.grid(True)
```

```python
selected_columns = ['math score', 'reading score', 'writing score']
correlation_matrix = df[selected_columns].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title(f'Heatmap of Correlations - StudentID: {student_id} - {current_datetime}')
plt.show()
```
